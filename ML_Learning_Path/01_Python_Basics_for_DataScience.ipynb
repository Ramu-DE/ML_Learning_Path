## 1. Variables & Data Types

```python
# Python has dynamic typing — no need to declare types
name    = "Alice"          # str
age     = 28               # int
height  = 5.6              # float
is_data_scientist = True   # bool

print(f"Name: {name}, Age: {age}, Height: {height}")
print(f"Types: {type(name)}, {type(age)}, {type(height)}")
```

```python
# Type conversion
x = "42"
print(int(x) + 8)      # 50 — string → int
print(float(x) * 1.5)  # 63.0
print(str(100) + " items")
```

## 2. Collections

```python
# List — ordered, mutable
fruits = ["apple", "banana", "cherry", "mango"]
print(fruits[0])       # first item
print(fruits[-1])      # last item
print(fruits[1:3])     # slice: ['banana', 'cherry']

fruits.append("orange")
fruits.remove("banana")
print(fruits)
```

```python
# Dictionary — key-value store (very common in data science)
student = {
    "name": "Bob",
    "grades": [88, 92, 79, 95],
    "enrolled": True
}

print(student["name"])
print(student.get("age", "N/A"))   # safe get with default

# Loop over key-value pairs
for key, val in student.items():
    print(f"  {key}: {val}")
```

```python
# Set — unique values only (great for deduplication)
scores = [10, 20, 20, 30, 10, 40]
unique_scores = set(scores)
print("Unique:", unique_scores)
print("Count of unique:", len(unique_scores))
```

## 3. Control Flow

```python
# if / elif / else
temperature = 35

if temperature > 40:
    print("Extreme heat")
elif temperature > 30:
    print("Hot day — stay hydrated!")
elif temperature > 20:
    print("Pleasant weather")
else:
    print("Cold")
```

```python
# for loop with enumerate
cities = ["Mumbai", "Delhi", "Bangalore", "Chennai"]

for i, city in enumerate(cities):
    print(f"{i+1}. {city}")
```

```python
# while loop with a real-world feel
balance = 1000
month = 0
while balance < 2000:
    balance *= 1.05   # 5% monthly growth
    month += 1

print(f"It takes {month} months to double to ₹{balance:.2f}")
```

## 4. Functions

```python
def calculate_bmi(weight_kg, height_m):
    """Calculate Body Mass Index."""
    bmi = weight_kg / (height_m ** 2)
    if bmi < 18.5:
        category = "Underweight"
    elif bmi < 25:
        category = "Normal"
    elif bmi < 30:
        category = "Overweight"
    else:
        category = "Obese"
    return round(bmi, 2), category

bmi, cat = calculate_bmi(70, 1.75)
print(f"BMI: {bmi} → {cat}")
```

```python
# Lambda functions — small anonymous functions
square  = lambda x: x ** 2
add     = lambda x, y: x + y

nums = [3, 1, 4, 1, 5, 9, 2, 6]
print("Sorted:", sorted(nums))
print("Sorted by square:", sorted(nums, key=lambda x: x**2, reverse=True))
```

## 5. List Comprehensions — Pythonic Power

```python
# Traditional loop vs comprehension
numbers = range(1, 11)

# Traditional
squares_loop = []
for n in numbers:
    if n % 2 == 0:
        squares_loop.append(n**2)

# Comprehension — one clean line
squares_comp = [n**2 for n in numbers if n % 2 == 0]

print("Even squares:", squares_comp)

# Dictionary comprehension
word_lengths = {word: len(word) for word in ["data", "science", "python", "machine", "learning"]}
print(word_lengths)
```

## 6. File I/O

```python
import os, csv, json

# Write a CSV file
data = [["Name","Age","Score"], ["Alice",25,88], ["Bob",30,92], ["Carol",22,79]]

with open("/tmp/sample.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(data)

# Read it back
with open("/tmp/sample.csv") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

```python
# JSON — the format of APIs and configs
config = {"model": "RandomForest", "n_estimators": 100, "max_depth": 5}

with open("/tmp/config.json", "w") as f:
    json.dump(config, f, indent=2)

with open("/tmp/config.json") as f:
    loaded = json.load(f)

print(loaded)
```

## 7. Error Handling

```python
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Error: cannot divide by zero")
        return None
    except TypeError as e:
        print(f"Type error: {e}")
        return None
    else:
        return result
    finally:
        print(f"Attempted: {a} / {b}")

print(safe_divide(10, 2))
print(safe_divide(10, 0))
```

## 8. Practice Exercises

```python
# Exercise 1: Write a function that takes a list of numbers
# and returns (min, max, mean, median) without using any library.

def stats(nums):
    nums_sorted = sorted(nums)
    n = len(nums_sorted)
    minimum = nums_sorted[0]
    maximum = nums_sorted[-1]
    mean    = sum(nums_sorted) / n
    if n % 2 == 0:
        median = (nums_sorted[n//2 - 1] + nums_sorted[n//2]) / 2
    else:
        median = nums_sorted[n//2]
    return minimum, maximum, round(mean, 2), median

data = [23, 7, 45, 12, 67, 34, 89, 5, 56, 28]
mn, mx, mean, med = stats(data)
print(f"Min={mn}, Max={mx}, Mean={mean}, Median={med}")
```

```python
# Exercise 2: Word frequency counter (real-world NLP warm-up)
text = """
Data science is the field of study that combines domain expertise programming skills
and knowledge of mathematics and statistics to extract meaningful insights from data
Data science practitioners apply machine learning algorithms to numbers text images
video audio and more to produce artificial intelligence AI systems to perform tasks
that ordinarily require human intelligence
"""

words = text.lower().split()
freq  = {}
for w in words:
    freq[w] = freq.get(w, 0) + 1

top10 = sorted(freq.items(), key=lambda x: x[1], reverse=True)[:10]
print("Top 10 words:")
for word, count in top10:
    print(f"  '{word}': {count}")
```
